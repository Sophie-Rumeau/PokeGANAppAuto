```python
from pathlib import Path
import sys

ROOT = Path('.').resolve()
SRC_DIR = ROOT / 'src'
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from pokestats.preprocessing import preprocess_dataset
from pokestats.ctgan_pipeline import train_and_sample_ctgan
```

```python
pre = preprocess_dataset(
    input_csv=ROOT / 'pokemon_data.csv',
    out_dir=ROOT / 'processed',
)

print('Preprocessing done')
print('-', pre['full_out'])
print('-', pre['ctgan_out'])
print('-', pre['meta_out'])
pre['ctgan_df'].head(3)
```

```python
result = train_and_sample_ctgan(
    train_csv=ROOT / 'processed/pokemon_ctgan_train.csv',
    artifacts_dir=ROOT / 'artifacts',
    epochs=300,
)

print('CTGAN done')
print('-', result['model_path'])
print('-', result['synth_path'])
print('-', result['synth_cond_path'])
if not result['eval_df'].empty:
    display(result['eval_df'])
```
